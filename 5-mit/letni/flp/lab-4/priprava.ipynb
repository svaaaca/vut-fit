## Syntaxe jazyka Rust

Rust syntakticky připomíná C/C++ (bloky ve `{}`, `;`, operátory), ale přidává:
- přísnější typový systém a inferenci typů,
- vlastnictví (ownership), půjčování (borrowing) a lifetimes,
- pattern matching a "algebraické" typy (`enum`),
- idiom "výrazy vrací hodnotu".

````rust
fn main() {
    let x = 3 + 4;
    if x > 5 {
        println!("x je vetsi nez 5");
    }
}
main();
````

## Bloky a středníky: výraz vs. příkaz

V Rustu je většina konstrukcí **výraz** (*expression*) – tedy něco, co může mít hodnotu.

- Blok `{ ... }` může **vrátit hodnotu**.
- **Poslední řádek bloku bez středníku `;`** je návratová hodnota bloku.
- Pokud na konec dáme `;`, z výrazu se stane **příkaz** (*statement*) a hodnota se „zahodí“
– výsledkem bloku pak bývá prázdná n-tice `()`.

Pozn.: `return` existuje, ale idiomatické je často vracet hodnotu právě posledním výrazem.


````rust
fn main() {
    let a = {
        let x = 10;
        x + 5   // bez ; => vrací 15
    };
    let b = {
        let x = 10;
        x + 5;  // se ; => vrací ()
    };
    println!("{a} {:?}", b);
}
main();
````

## Proměnné jsou defaultně neměnné (immutable)

V Rustu jsou proměnné vytvořené pomocí `let` **defaultně neměnné** (*immutable*). Jejich hodnotu nemůžeme změnit.

- Změny musíme **explicitně** povolit pomocí `mut` (`let mut x = ...`).
- Neměnnost zlepšuje čitelnost a bezpečnost: vidíme `let x = ...` a víme, že se `x` už nezmění.
- Kompilátor díky tomu snadněji odhalí chyby (nechtěné přepisování hodnot) a lépe optimalizuje.
- Oproti C/Pythonu Rust vede k tomu, aby byla mutabilita spíš výjimka než pravidlo.

**POZOR:**
- `mut` říká, že *proměnná* může ukazovat na jinou hodnotu / být přepsaná.
- To neznamená automaticky, že i *data uvnitř* jsou vždy měnitelná!


````rust
fn main() {
    let x = 10;
    let mut y = 10;
    y += 1;
    println!("{x} {y}");
}
main();
````

## Typy: statické, ale s inferencí

Rust je **staticky typovaný** jazyk: typy všech výrazů musí být známé už **při kompilaci**. Díky tomu kompilátor odhalí řadu chyb ještě před spuštěním.

Zároveň má Rust silnou **inferenci typů**:
- Typ často nemusíme psát, kompilátor ho odvodí z kontextu (např. z přiřazení, použitých operací, volání funkcí).
- Typy se obvykle dopisují jen tehdy, když je kontext nejednoznačný nebo chceš být explicitní (čitelnost, API, hranice mezi moduly).

Praktické důsledky:
- Kód bývá stručný (díky inferenci), ale stále bezpečný (díky statickému typování).
- Když se typ nedá spolehlivě odvodit, kompilátor nás donutí ho doplnit.


````rust
fn main() {
    let a = 123;      // i32 (typicky)
    let b: u64 = 123; // explicitně
    println!("{a} {b}");
}
main();
````

### **PŘÍKLAD 1:** Práce s datovými typy

````rust
// Doplňovací cvičení: proměnné a datové typy (+ assert)
// Řešení je už přímo vepsané, ale označené markerem pro automatické odstranění.

fn main() {
    // 1) a = Cele cislo se znamenkem (typu i32) s hodnotou -15
    let a: i32 = -15 as i32;
    println!("a = {a}");

    // 2) b = Cele cislo bez znamenka (typu u32) s hodnotou 100
    let b: u32 = 100 as u32;
    println!("b = {b}");

    // 3) c = Desetinne cislo (typu f64) s hodnotou 3.5
    let c: f64 = 3.5 as f64;
    println!("c = {c}");

    // 4) ok = Pravdivostni hodnota (typu bool) s hodnotou true
    let ok: bool = true as bool;
    println!("ok = {ok}");

    // 5) ch = Jeden znak (char) s hodnotou 'R'
    let ch: char = 'R' as char;
    println!("ch = {ch}");

    // 6) text = Text (String) s hodnotou "Ahoj"
    let text: String = String::from("Ahoj");
    println!("text = {text}");

    // 7) u = Cele cislo (i32) s hodnotou 40, kterou pak zvysime o 2
    let mut u: i32 = 40 as i32;
    println!("u (start) = {u}");
    // Inkrementace u o hodnotu 2:
    u += 2;
    println!("u (after +2) = {u}");

    // ================================
    // Dalsi radky uz NEUPRAVUJTE!
    assert_eq!(a, -15);
    assert_eq!(b, 100);
    assert_eq!(c, 3.5);
    assert_eq!(ok, true);
    assert_eq!(ch, 'R');
    assert_eq!(text, "Ahoj");
    assert_eq!(u, 42);
}
main();
````

## Řídicí struktury

### `if / else`

V Rustu je `if` podobné jako v C, ale s několika důležitými rozdíly:

- Podmínka musí být typu `bool` (nejde používat "pravdivost" čísel jako v C, tj. `if 1 {}` je chyba).
- `if` je **výraz** (*expression*), takže může vracet hodnotu:
  - Výsledek `if` tak můžeme přiřadit do proměnné.
  - Všechny větve (`if`, `else if`, `else`) musí vracet **stejný typ**.
- Závorky kolem podmínky nejsou povinné: píše se `if condition { ... }`.
- Tělo `if/else` je blok v `{}` a může obsahovat libovolný kód.


````rust
fn main() {
    let n = 7;

    // klasicke vetveni
    if n % 2 == 0 {
        println!("{n} je sude");
    } else {
        println!("{n} je liche");
    }

    // if jako vyraz (vraci hodnotu)
    let popis = if n > 10 { "velke" } else { "male" };
    println!("n je {popis}");

    // vice vetvi
    let znamka = 2;
    let text = if znamka == 1 {
        "vyborne"
    } else if znamka == 2 {
        "chvalitebne"
    } else {
        "jinak"
    };
    println!("Hodnoceni: {text}");
}
main();
````

### Cyklus `loop`

- `loop` běží "navždy", dokud jej explicitně neukončíme (`break`).
- Přeskočení na další iteraci: `continue`.
- `loop` je **výraz**, takže může vracet hodnotu: `let x = loop { ... break hodnota; };`

````rust
fn main() {
    let mut i = 0;

    loop {
        i += 1;

        if i == 2 {
            continue;
        }

        println!("i = {i}");

        if i == 4 {
            break;
        }
    }

    let result = loop {
        let x = 5;
        break x * 2;
    };
    println!("result = {result}");
}
main();
````

### Cyklus `while`
- `while` opakuje blok, dokud platí podmínka.
- Podmínka musí být typu `bool`.
- Smyčka se vyhodnocuje "předem": když je podmínka hned na začátku `false`, tělo se vůbec neprovede.
- Uvnitř lze použít `break` pro ukončení a `continue` pro přeskočení zbytku iterace.

````rust
fn main() {
    let mut n = 5;

    while n > 0 {
        println!("n = {n}");
        n -= 1;
    }

    println!("done");
}
main();
````

### Cyklus `for`
- `for` iteruje přes **iterátor** (nejčastěji rozsah nebo kolekci).
- Rozsah: `0..n` (bez poslední hodnoty), nebo `0..=n` (včetně).
- `for` je bezpečný a idiomatický: neřešíme ručně indexy ani podmínky ukončení.
- U kolekcí se typicky iteruje přes `iter()` (čtení), `iter_mut()` (změna prvků), nebo `into_iter()` (přesun vlastnictví).
- Stejně jako u jiných cyklů lze použít `break` a `continue`.

````rust
fn main() {
    for i in 0..3 {
        println!("i = {i}");
    }

    let v = vec![10, 20, 30];
    for x in v.iter() {
        println!("x = {x}");
    }
}
main();
````

### Labelované cykly

Rust umožňuje pojmenovat smyčku pomocí **štítku** (label) ve tvaru `'jmeno:` před smyčkou.

- Užitečné hlavně u **vnořených cyklů**, kde chceme řídit tok cyklu vnějšího
- `break 'label;` ukončí konkrétně označenou smyčku.
- `continue 'label;` přeskočí na další iteraci konkrétně označené smyčky.
- Štítky fungují pro `loop`, `while` i `for`.
- Umožňují vyhnout se pomocným proměnným typu "flag", nebo složitým podmínkám.

````rust
fn main() {
    'outer: for i in 0..3 {
        for j in 0..3 {
            if i == 1 && j == 1 {
                println!("break outer at i={i}, j={j}");
                break 'outer;
            }

            if j == 2 {
                println!("continue outer at i={i}, j={j}");
                continue 'outer;
            }

            println!("i={i}, j={j}");
        }
    }

    println!("done");
}
main();
````

### **PŘÍKLAD 2:** Jednoduchá práce s cykly
- Představte si tabulku s **řádky** `r = 1, 2, 3, 4` a **sloupci** `c = 1, 2, 3`.
- V každé buňce je hodnota `v = r * c`.
- Např. v buňce na 2. řádku a 3. sloupci je hodnota `2 * 3 = 6`

Na vyznačené místo **doplňte kód**, který:
- Projde všechny kombinace `(r, c)`
- Spočítá, kolik buněk zpracoval a uloží je do proměnné `count`
- Sudé hodnoty sečte a výsledek uloží `sum_even`

Do kódu mimo vyznačené místo **nezasahujte**!

````rust
fn main() {
    let mut sum_even: i32 = 0;
    let mut count: i32 = 0;

    for r in 1..=4 {
        for c in 1..=3 {
            count += 1;
            let rc = r * c;
            if rc % 2 == 0 {
                sum_even += rc;
            }
        }
    }

    println!("count = {count}, sum_even = {sum_even}");

    // --- Kontrola (neupravujte) ---
    assert_eq!(count, 12);      // 4 * 3 bunek
    assert_eq!(sum_even, 44);   // sude hodnoty z nasobici tabulky
}
main();
````

## Funkce, parametry a návratové typy

Funkce v Rustu se definují pomocí `fn`. Každý parametr má uvedený typ a (když funkce něco vrací) uvádí se i **návratový typ** za šipkou `->`.

- Parametry: `jméno: Typ` (typ je povinný).
- Návratový typ: `-> Typ` (pokud chybí, funkce vrací jednotkový typ `()` - prázdná n-tice).
- Návrat hodnoty:
  - idiomaticky se vrací **posledním výrazem bez středníku**,
  - `return` existuje, ale používá se spíš výjimečně (např. pro předčasný návrat).

````rust
fn add(a: i32, b: i32) -> i32 {
    a + b
}

fn main() {
    println!("{}", add(2, 3));
}
main();
````

## Ownership (vlastnictví)

Rust nepoužívá garbage collector (jako Python, Java) ani nás nenutí spravovat paměť ručně (jako C/C++). Místo toho má pravidla **vlastnictví (ownership)**, která kompilátor vynucuje už při překladu.

Základní principy:
- **Každá hodnota má právě jednoho vlastníka** (proměnnou/vektor/strukturu/..., která ji "vlastní").
- Když vlastník zanikne (např. konec bloku/funkce), hodnota se automaticky **uvolní** (*drop*).
- Přesun vlastnictví (**move**) je výchozí chování pro typy, které vlastní data na haldě (např. `String`, `Vec<T>`):
  - při přiřazení nebo předání do funkce se hodnota typicky "přesune"
  - původní proměnná se pak už nesmí použít.
- U typů s vlastností `Copy` (např. číselné typy, `bool`, `char`) se místo přesunu udělá **kopie**.

Místo "předám pointer a uvidíme" se v Rustu jasně vyjadřuje, **kdo data vlastní** a **kdo si je jen půjčuje.**

````rust
fn main() {
    let s1 = String::from("ahoj");
    let s2 = s1;           // move
    // println!("{s1}");   // chyba: s1 už není platné
    println!("{s2}");
}
main();
````

## Projekty v Cargo

Cargo je **oficiální** nástroj Rustu pro správu projektů a balíčků (packages). Používá se k vytváření nových projektů, překladu programu, spouštění testů, správě závislostí a dalším běžným úkolům při vývoji. 

### Terminologie

- **Crate** - Základní jednotka kompilace. Může to být buď knihovna (library crate), nebo spustitelný program (binary crate).
- **Package** - Projekt spravovaný systémem Cargo. Obsahuje soubor `Cargo.toml` a **jednu nebo více crates**. Přesněji: libovolný počet binárních crates, ale nejvýše jednu library crate.

### Spustitelná binárka (binary crate)

**Spustitelná binárka** (binary crate, bin crate) je hotový program, který se dá přímo spustit v operačním systému. Musí mít vstupní bod `fn main()` a v Cargo projektu bývá typicky v souboru `src/main.rs` (případně lze mít více binárek v `src/bin/*.rs`). Po překladu vznikne **spustitelný soubor** v `target/debug/` nebo `target/release/`, který je určený k **běhu**.

Spustitelná binárka obsahuje **strojový kód** pro konkrétní platformu - je zkompilovaná pro určitou **architekturu CPU** (např. `x86_64`, `aarch64`) a pro konkrétní **OS a formát binárky** (např. ELF na Linuxu, PE na Windows, Mach-O na macOS). Binárka může být staticky nebo dynamicky linkovaná se systémovými knihovnami (např. libc), a proto obvykle nejde jen tak přenést mezi různými OS/architekturami bez překompilování


`cargo new prestiz` vytvoří **nový package** s následující adresářovou strukturou:

```
prestiz/
├── Cargo.toml
└── src/
    └── main.rs
```

- **Cargo.toml** - Manifest balíčku (package): název `prestiz`, verze, *edition*, (později) závislosti.
- **src/main.rs** - Kód vstupního bodu programu (**bin crate**) – obsahuje `fn main()`.

### Knihovna (library crate)

**Knihovna** (library crate, lib crate) je balík znovupoužitelného kódu, který se sám o sobě nespouští. Nemá `main`, ale poskytuje veřejné funkce, typy a moduly přes `pub`, přičemž kořen knihovny je obvykle `src/lib.rs`. Knihovna se používá tak, že ji jiný projekt přidá jako závislost do `Cargo.toml` (nebo jako `path` dependency ve workspace) a pak její API importuje pomocí `use`.

Výsledkem překladu knihovny není samostatná aplikace, ale **knihovní artefakt**, který se pak linkuje do binárky nebo jiné knihovny. V Rustu je to nejčastěji `*.rlib` (Rustová "static" knihovna), případně můžeme cílit i na dynamickou knihovnu nebo C-ABI podle toho, jak nastavíme typ crate (např. `cdylib`, `staticlib`). Knihovna typicky obsahuje zkompilovaný kód a metadata tak, aby ji mohl linker spojit s výsledným programem.


`cargo new --lib prestiz` vytvoří následující:

```
prestiz/
├── Cargo.toml
└── src/
    └── lib.rs
```

- **Cargo.toml** - Manifest balíčku (package): název `prestiz`, verze, *edition*, (později) závislosti.
- **src/lib.rs** - Kořen knihovny (**lib crate**) – definuje veřejné API (exportované přes `pub`).

> Pozn.: Knihovna nemá `fn main()`. Typicky ji používá jiný crate jako závislost, nebo ji kombinujeme s binárkou (v jednom package může být současně `src/lib.rs` i `src/main.rs`).

> Pozn. 2: V jednom Cargo balíčku můžeš mít klidně obojí zároveň: `src/lib.rs` pro sdílenou logiku (knihovnu) a `src/main.rs` jako spouštěcí "wrapper" (binárku), která tu logiku používá. To je častý a praktický styl, protože logika je lépe testovatelná v knihovně a binárka pak řeší hlavně spouštění a práci s parametry.
 

### Překlad:
- `cargo build` vytvoří artefakty v `target/debug/`
  - **Spustitelná binárka:** `target/debug/prestiz`
  - **Knihovna:** typicky `target/debug/deps/libprestiz-<hash>.rlib`
  - Překlad pro **ladění** / **vývoj** (profil `dev`).
  - Kompiluje **rychleji** (kratší časy kompilace).
  - Výsledná binárka/knihovna bývá **pomalejší** po nalinkování (má méně optimalizací).
  - Standardně obsahuje **ladící informace** (pro debugger, stacktrace, lepší hlášky).
  - Typicky má zapnuté **runtime kontroly** (např. kontroly přetečení v `debug` módu u celočíselných operací – obvykle končí panikou).

Pro release:
- `cargo build --release` vytvoří artefakty v `target/release/`
  - **Spustitelná binárka:** `target/release/prestiz`
  - **Knihovna:** typicky `target/release/deps/libprestiz-<hash>.rlib`
  - Překlad pro **produkci** (profil `release`).
  - Kompiluje **pomaleji** (protože provádí optimalizace).
  - Výsledná binárka/knihovna je obvykle **výrazně rychlejší** a často i **menší**.
  - Standardně mívá **méně ladících informací** (dá se ale zapnout v `Cargo.toml`).
  - Chování overflow u celočísel: v release profilu se typicky **nepanikuje**, ale používá se "wrapping" (přetečení se přetočí), pokud explicitně nepoužijeme kontrolované metody (`checked_*`, `overflowing_*`, atd.) nebo nenastavíme kontroly.

Profily se dají upravit v `Cargo.toml`, např.:
```toml
[profile.dev]
opt-level = 0

[profile.release]
opt-level = 3
```

### Spuštění:
- `cargo run`
  - Projekt **přeloží** (pokud je to potřeba) a pak **spustí** výslednou binárku.
  - Pokud už je vše aktuální, překlad se přeskočí a jen se spustí program.
  - Standardně spouští **debug** profil (tj. jako `cargo build`), takže binárka je v `target/debug/`.
  - Pro release režim se používá `cargo run --release`.

Spuštění bez ladících výpisů:
- `cargo run -q` (nebo `cargo run --quiet`)
  - Dělá totéž co `cargo run`, ale **potlačí většinu výpisů Cargo** (např. hlášky o kompilaci, "Compiling...", "Finished...").
  - Program samotný se spustí normálně a vypisuje se pouze jeho **vlastní výstup** (`println!` apod.).

### Další užitečné příkazy v Cargo
`cargo check`
- Projekt **zkontroluje a typově ověří**, ale nedělá finální krok generování strojového kódu.
- Proto je obvykle rychlejší než cargo build.
- Pozor, některé chyby se odhalí až při plné kompilaci!

`cargo test`
- Provede **testy** balíčku.
- Testy hledá uvnitř souborů v `src` a také v adresáři `tests/`.
- V `src` jsou typicky unit testy a dokumentační testy, zatímco `tests/` slouží pro integrační testy.

`cargo doc`
- K aktuálnímu package **vygeneruje dokumentaci** ve formě HTML souborů.
- Zároveň generuje dokumentaci i pro **závislosti**. Lze vypnout přepínačem `--no-deps`.
- Dokumentaci ukládá do adresáře `target/doc`.

`cargo fmt`
- Automaticky **zformátuje zdrojový kód** podle konvencí Rustu.
- Užitečné: před commitem, sjednocení formátu při více programátorech apod.
- Může být nutné doinstalovat: `rustup component add rustfmt`

`cargo clippy`
- Detekuje (a případně opravuje) chyby v kódu.
- Kontroluje kvalitu kódu.
- Upozorňuje na podezřelé konstrukce, neidiomatický styl.
- Ne všechna doporučení je nutné přijmout.
- Má více režimů, nejpřísnější jsou `pedantic` a `restriction`.
- Může být nutné doinstalovat: `rustup component add clippy`

`cargo publish`
- **Publikuje** package do registru (standardně `crates.io`).
- Pro tyto účely vytváří distribuovatelný archiv `.crate` (byť jde o package).
- Před publikací provádí kontroly (závislosti, přeložitelnost, ..). Chceme-li jen kontrolu, tak `--dry-run`.
