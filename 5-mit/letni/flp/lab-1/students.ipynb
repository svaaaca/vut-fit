**Příklad 1:** Napište *výrazy*, které budou manipulovat s řetězci. Není nutné nijak generalizovat, psát funkce, opravdu jen triviální řešení s vestavěnými funkcemi, o kterých jsme mluvili.

```haskell
-- příklad)
-- dostane hodnotu
"Milujeme FLP"
-- vrátí
"me FLP"
-- očekávané řešení:
drop 6 "Milujeme FLP"

-- a)
-- dostane hodnotu
"Curry is awesome"
-- vrátí
"Curry is awesome!"
-- b)
-- dostane hodnotu
"Curry is awesome!"
-- vrátí
"y"
-- c)
-- dostane hodnotu
"Curry is awesome!"
-- vrátí
"awesome!"
```

```haskell
-- EX01a
"Curry is awesome" ++ "!"

-- EX01b
"Curry is awesome" !! 4

-- EX01c
drop 9 "Curry is awesome!"
```

**Příklad 2:** Napište funkci `thirdLetter`, která dostane libovolný řetězec (`String`) a vrátí jeho třetí znak (`Char`). Nezapomeňte na typovou signaturu. (Neřešte žádné kontroly délky řetězce.)

```haskell
-- EX02
thirdLetter :: String -> Char
thirdLetter x = x !! 2
```

**Příklad 3:** Napište funkci (vč. typové anotace!), která řekne, zda je daný seznam palindrom (stejný odpředu i odzadu). Využijte k tomu vestavěnou funkci `reverse`.

```haskell
:t reverse
reverse "ahoj"
```

```haskell
-- EX03
isPalindrome :: Eq a => [a] -> Bool
isPalindrome x = x == reverse x
```

**Příklad 4:** S využitím konstrukce if-then-else napište funkci `myAbs`, která vrátí absolutní hodnotu čísla.

```haskell
-- EX04
myAbs :: (Num a, Ord a) => a -> a
myAbs x = if x < 0 then -x else x
```

**Příklad 5:** Podle typové anotace doplňte funkci `f`. Využijte funkce `fst`/`snd`, které vrací první/druhý prvek z dvojice.

*Možná je to na první pohled zvláštní, ale na základě takhle „vágní“ typové anotace opravdu existuje právě jedna možnost, jak tohle udělat správně.*

```haskell
:t fst
:t snd
```

```haskell
-- EX05
f :: (a, b) -> (c, d) -> ((d, b), (a, c))
f x y = ((snd y, snd x), (fst x, fst y))
```
